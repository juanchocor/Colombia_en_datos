```python
from pathlib import Path
import pandas as pd
from IPython.display import display

PROJECT_DIR = Path.cwd().resolve().parent
FILE_PATH = PROJECT_DIR / "episodios" / "003_d1_vs_tiendas" / "data" / "raw" / "Registro_Maestro_Retail_Descuento_Duro (1).xlsx"

workbook = pd.ExcelFile(candidates[0])
SHEETS = workbook.sheet_names

print(f'Archivo: {FILE_PATH.name}')
print(f'Hojas encontradas: {len(SHEETS)}')
print(SHEETS)
```

# Inventario de hojas

```python
inventory = []

for sheet_name in SHEETS:
    sheet = pd.read_excel(candidates[0], sheet_name=sheet_name, header=None)
    first_values = sheet.stack().head(3).astype(str).tolist()
    inventory.append({
        'hoja': sheet_name,
        'filas': sheet.shape[0],
        'columnas': sheet.shape[1],
        'celdas_no_vacias': int(sheet.notna().sum().sum()),
        'vista_previa': ' | '.join([str(v) if pd.notna(v) else '' for v in first_values]),
    })

inventory_df = pd.DataFrame(inventory)
display(inventory_df)
```

# Explorando 1 hoja

```python
SHEET_TO_EXPLORE = 'Datos_D1_Serie_2019_2025'

if SHEET_TO_EXPLORE not in SHEETS:
    raise ValueError(f'La hoja {SHEET_TO_EXPLORE!r} no existe. Opciones: {SHEETS}')

df = pd.read_excel(candidates[0], sheet_name=SHEET_TO_EXPLORE, header=None)

print(f'Hoja: {SHEET_TO_EXPLORE}')
print(f'df.shape: {df.shape}')
print('\ndf.columns:')
print(df.columns.tolist())
print('\ndf.dtypes:')
display(df.dtypes.to_frame(name='dtype'))
print('\ndf.head(15):')
display(df.head(15))
print('\ndf.info():')
df.info()

missingness = pd.DataFrame({
    'valores_faltantes': df.isna().sum(),
    'porcentaje_faltante': (df.isna().mean() * 100).round(2),
})
print('\ndf.isna().sum():')
display(missingness)
```
