# Exploración inicial de fuentes

Este notebook revisa cada archivo CSV disponible en `data/raw/`. Su propósito es documentar de manera reproducible la estructura, los tipos de datos, una muestra inicial y los valores faltantes de cada fuente.

## 1. Configuración

Los datos originales se leen desde una ruta relativa al proyecto. No se modifican los archivos de origen.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

PROJECT_DIR = Path.cwd().resolve().parent
RAW_DIR = PROJECT_DIR / 'data' / 'raw'
CSV_FILES = sorted(RAW_DIR.glob('*.csv'))

print(f'Carpeta de datos: {RAW_DIR}')
print(f'Archivos CSV encontrados: {len(CSV_FILES)}')
[file.name for file in CSV_FILES]
```

## 2. Función de carga

Se prueban las codificaciones más frecuentes en las fuentes disponibles. `sep=None` permite a pandas inferir el separador.

```python
def read_csv_with_fallback(file_path: Path) -> tuple[pd.DataFrame, str]:
    """Carga un CSV probando codificaciones frecuentes."""
    last_error = None
    for encoding in ('utf-8', 'latin-1'):
        try:
            dataframe = pd.read_csv(file_path, encoding=encoding, sep=None, engine='python')
            return dataframe, encoding
        except (UnicodeDecodeError, pd.errors.ParserError) as error:
            last_error = error
    raise last_error
```

## 3. Exploración por archivo

Para cada archivo se muestran: `df.shape`, `df.columns`, `df.dtypes`, `df.head()`, `df.info()` y `df.isna().sum()`.

```python
if not CSV_FILES:
    raise FileNotFoundError(f'No se encontraron archivos CSV en {RAW_DIR}')

for file_path in CSV_FILES:
    df, encoding = read_csv_with_fallback(file_path)

    print('═' * 100)
    print(f'ARCHIVO: {file_path.name}')
    print(f'Codificación usada: {encoding}')

    print('\n1. df.shape')
    print(df.shape)

    print('\n2. df.columns')
    print(df.columns.tolist())

    print('\n3. df.dtypes')
    display(df.dtypes.to_frame(name='dtype'))

    print('\n4. df.head()')
    display(df.head())

    print('\n5. df.info()')
    df.info()

    print('\n6. df.isna().sum()')
    display(df.isna().sum().to_frame(name='valores_faltantes'))
```
