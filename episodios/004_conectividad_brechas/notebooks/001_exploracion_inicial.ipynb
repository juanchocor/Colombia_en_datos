# primera gráfica

Evolución de los 6 estratos. 

```python
estrato_q4 = (
    residenciales_q4
    .groupby(["ANNO", "SEGMENTO"], as_index=False)["ACCESOS"]
    .sum()
)

estrato_pivot = (
    estrato_q4
    .pivot(index="ANNO", columns="SEGMENTO", values="ACCESOS")
)

estrato_pivot
```

calculamos el cambio 2017 → 2025

```python
cambio_estrato = pd.DataFrame({
    "accesos_2017": estrato_pivot.loc[2017],
    "accesos_2025": estrato_pivot.loc[2025]
})

cambio_estrato["cambio_absoluto"] = (
    cambio_estrato["accesos_2025"] -
    cambio_estrato["accesos_2017"]
)

cambio_estrato["cambio_%"] = (
    cambio_estrato["accesos_2025"] /
    cambio_estrato["accesos_2017"] - 1
) * 100

cambio_estrato.sort_values("cambio_%", ascending=False)
```

Pregunta 2
¿Dónde está la brecha territorial?

```python
municipal_q4 = (
    residenciales_q4
    .groupby(
        [
            "ID_MUNICIPIO",
            "MUNICIPIO",
            "ID_DEPARTAMENTO",
            "DEPARTAMENTO",
            "ANNO"
        ],
        as_index=False
    )["ACCESOS"]
    .sum()
)
```

```python
municipal_cambio = (
    municipal_q4
    .pivot_table(
        index=[
            "ID_MUNICIPIO",
            "MUNICIPIO",
            "ID_DEPARTAMENTO",
            "DEPARTAMENTO"
        ],
        columns="ANNO",
        values="ACCESOS",
        fill_value=0
    )
    .reset_index()
)
```

```python
municipal_cambio = municipal_cambio[
    [
        "ID_MUNICIPIO",
        "MUNICIPIO",
        "ID_DEPARTAMENTO",
        "DEPARTAMENTO",
        2017,
        2025
    ]
].copy()

municipal_cambio = municipal_cambio.rename(
    columns={
        2017: "ACCESOS_2017",
        2025: "ACCESOS_2025"
    }
)
```

```python
municipal_cambio["CAMBIO_ABSOLUTO"] = (
    municipal_cambio["ACCESOS_2025"] -
    municipal_cambio["ACCESOS_2017"]
)

municipal_cambio["CAMBIO_%"] = (
    (municipal_cambio["ACCESOS_2025"] /
     municipal_cambio["ACCESOS_2017"]) - 1
) * 100
```

Mayores crecimientos relativos

```python
municipal_cambio.sort_values(
    "CAMBIO_%",
    ascending=False
).head(20)
```

Mayores aumentos absolutos

```python
municipal_cambio.sort_values(
    "CAMBIO_ABSOLUTO",
    ascending=False
).head(20)
```

Municipios con una base mínima en 2017

```python
municipal_cambio[
    municipal_cambio["ACCESOS_2017"] >= 1000
].sort_values(
    "CAMBIO_%",
    ascending=False
).head(20)
```

Los municipios que más retrocedieron

```python
municipal_cambio.sort_values(
    "CAMBIO_%",
    ascending=True
).head(20)
```

```python
# Nos quedamos únicamente con Q4
df_q4 = df1[df1["TRIMESTRE"] == 4].copy()
```

```python
municipios_anual = (
    df_q4
    .groupby(
        [
            "ANNO",
            "ID_MUNICIPIO",
            "MUNICIPIO",
            "ID_DEPARTAMENTO",
            "DEPARTAMENTO"
        ],
        as_index=False
    )["ACCESOS"]
    .sum()
)
```

hacemos el pivot

```python
municipios_pivot = (
    municipios_anual
    .pivot_table(
        index=[
            "ID_MUNICIPIO",
            "MUNICIPIO",
            "ID_DEPARTAMENTO",
            "DEPARTAMENTO"
        ],
        columns="ANNO",
        values="ACCESOS",
        fill_value=0
    )
    .reset_index()
)
```

```python
municipios_pivot.columns.name = None
```

renombramos

```python
municipios_pivot = municipios_pivot.rename(
    columns={
        2017: "ACCESOS_2017",
        2025: "ACCESOS_2025"
    }
)
```

```python
ranking_municipios = municipios_pivot[
    (municipios_pivot["ACCESOS_2017"] > 0) &
    (municipios_pivot["ACCESOS_2025"] > 0)
].copy()
```

```python
ranking_municipios["CAMBIO_ABSOLUTO"] = (
    ranking_municipios["ACCESOS_2025"] -
    ranking_municipios["ACCESOS_2017"]
)

ranking_municipios["CAMBIO_%"] = (
    ranking_municipios["CAMBIO_ABSOLUTO"] /
    ranking_municipios["ACCESOS_2017"]
) * 100
```

```python
ranking_municipios.head()
```

Retomamos preguntas con los datos solidos despues de aplicar pivot.

Ranking por crecimiento porcentual

```python
top_crecimiento = (
    ranking_municipios
    .sort_values("CAMBIO_%", ascending=False)
    [[
        "MUNICIPIO",
        "DEPARTAMENTO",
        "ACCESOS_2017",
        "ACCESOS_2025",
        "CAMBIO_ABSOLUTO",
        "CAMBIO_%"
    ]]
    .head(20)
)

top_crecimiento
```

Ranking por crecimiento absoluto

```python
top_absoluto = (
    ranking_municipios
    .sort_values("CAMBIO_ABSOLUTO", ascending=False)
    [[
        "MUNICIPIO",
        "DEPARTAMENTO",
        "ACCESOS_2017",
        "ACCESOS_2025",
        "CAMBIO_ABSOLUTO",
        "CAMBIO_%"
    ]]
    .head(20)
)

top_absoluto
```
