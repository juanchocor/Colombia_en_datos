Normalizamos la fuente

```python
df_gini = df_gini.rename(columns={
    "DaneMun": "divipola",
    "Gini": "gini_movilidad"
})

df_gini["divipola"] = pd.to_numeric(
    df_gini["divipola"],
    errors="coerce"
).astype("Int64")

df_gini["gini_movilidad"] = pd.to_numeric(
    df_gini["gini_movilidad"],
    errors="coerce"
)
```

validamos que la fuente no tenga municipios repetidos

```python
print("Filas:", len(df_gini))
print("DIVIPOLA únicos:", df_gini["divipola"].nunique())
print("Duplicados DIVIPOLA:", df_gini["divipola"].duplicated().sum())
```

Creamos la variable de validez

```python
df_gini["gini_movilidad_valido"] = (
    df_gini["gini_movilidad"].between(0, 1)
)

# Los valores fuera de [0,1] pasan a faltante
df_gini.loc[
    ~df_gini["gini_movilidad_valido"],
    "gini_movilidad"
] = pd.NA
```

```python
display(
    df_gini[
        ["divipola", "Municipio", "gini_movilidad", "gini_movilidad_valido"]
    ].head(10)
)

print(
    "Valores inválidos:",
    (~df_gini["gini_movilidad_valido"]).sum()
)
```

Filtrar únicamente IPS

```python
ips_choco = salud_choco[
    (salud_choco["ClasePrestadorDesc"] == 
     "Instituciones Prestadoras de Servicios de Salud - IPS")
    &
    (salud_choco["en_universo_iot"])
].copy()

print("Sedes IPS dentro del universo:", len(ips_choco))
print("Municipios con IPS:", ips_choco["divipola_reps"].nunique())
```

```python
ips_municipio = (
    ips_choco
    .groupby("divipola_reps")
    ["CodigoHabilitacionSede"]
    .nunique()
    .reset_index(name="ips_total")
    .rename(columns={"divipola_reps": "divipola"})
)

ips_municipio = ips_municipio.sort_values(
    "ips_total",
    ascending=False
)

print(ips_municipio.to_string(index=False))
```

La definición metodológica:

ips_total: número de sedes de Instituciones Prestadoras de Servicios de Salud (IPS) registradas en el REPS para cada municipio del universo IOT, utilizando el código territorial de la sede (CodigoHabilitacionSede) y contando sedes únicas.

Fuente: REPS, corte 18 de agosto de 2026.
