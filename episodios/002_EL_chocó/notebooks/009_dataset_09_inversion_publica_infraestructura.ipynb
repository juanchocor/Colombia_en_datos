# Exploración y auditoría inicial de fuentes
## Índice de Oportunidades Territoriales (IOT) — Chocó

---

## 1. Objetivo de la exploración

Esta etapa tiene como objetivo realizar la exploración, identificación, limpieza inicial y evaluación de disponibilidad de las fuentes utilizadas para construir el Índice de Oportunidades Territoriales (IOT) del departamento del Chocó.

La exploración inicial no busca establecer relaciones causales, construir el índice ni interpretar los resultados territoriales. Su propósito es determinar qué información contienen las fuentes, cuál es su unidad de observación, qué tan bien pueden ser homologadas al universo municipal del IOT y cuáles pueden convertirse posteriormente en variables comparables.

El universo territorial de referencia está compuesto por **31 municipios del departamento del Chocó**.

Esta etapa constituye la frontera entre la exploración de las fuentes y la fase posterior de análisis metodológico del IOT.

---

## 2. Universo territorial

El universo territorial utilizado durante toda la construcción del proyecto corresponde al universo municipal definido para el IOT.

**Características del universo:**

- Departamento: Chocó
- Municipios: 31
- DIVIPOLA únicas: 31
- Unidad territorial objetivo: municipio

La DIVIPOLA constituye el identificador territorial preferente para integrar las diferentes fuentes.

Los nombres municipales homologados al universo IOT se utilizan como referencia descriptiva, mientras que los nombres originales de las fuentes se conservan cuando resulta relevante para garantizar trazabilidad.

---

## 3. Inventario general de las fuentes

Durante la exploración inicial se trabajó con diez datasets correspondientes a diferentes dimensiones de información territorial.

Los datasets fueron explorados individualmente antes de ser considerados para la construcción del IOT.

La clasificación general de cada fuente debe considerar:

- tema
- fuente original
- unidad de observación
- nivel territorial
- cobertura municipal
- temporalidad
- naturaleza de la información
- disponibilidad de DIVIPOLA
- necesidad de homologación
- posibilidad de agregación
- existencia de valores faltantes
- limitaciones metodológicas
- posibilidad de construir variables comparables

El inventario consolidado de los diez datasets se documenta a partir de los archivos de metodología, auditoría y cobertura generados durante la construcción de cada dataset.

---

## 4. Unidad de observación

Una de las primeras decisiones de la exploración fue identificar qué representa una fila en cada fuente.

La unidad de observación no se asumió automáticamente como municipal.

Dependiendo de la fuente, una fila puede representar:

- un municipio
- un municipio-año
- una institución
- una sede
- un evento
- un contrato
- un registro administrativo
- un documento
- un caso
- u otra unidad específica de la fuente.

Esta distinción es fundamental porque una fuente con múltiples registros por municipio no puede agregarse automáticamente sin comprender primero qué representa cada registro.

Por esta razón, antes de construir variables municipales se evaluó la unidad de observación de cada fuente.

---

## 5. Homologación territorial

La homologación territorial constituye una regla transversal del proyecto.

La prioridad utilizada es:

1. DIVIPOLA
2. código municipal oficial
3. combinación de departamento y municipio
4. nombre municipal
5. otros identificadores territoriales explícitos de la fuente

Cuando una fuente contiene DIVIPOLA, esta se utiliza como identificador principal.

Cuando no existe DIVIPOLA, se realiza homologación mediante nombres municipales u otros identificadores disponibles, conservando el valor original de la fuente.

Cuando resulta necesario normalizar texto para realizar la homologación, se pueden considerar diferencias como:

- mayúsculas y minúsculas
- tildes
- espacios
- abreviaciones
- nombres con paréntesis
- nombres de cabeceras municipales
- variaciones ortográficas

Sin embargo, la normalización utilizada para comparar nombres no debe alterar el valor original de la fuente.

No se considera que una mención territorial sea equivalente automáticamente a una observación municipal.

Asimismo, no se utilizan coincidencias difusas cuando puedan generar falsos positivos.

Toda homologación territorial relevante debe poder ser rastreada hasta el valor original de la fuente.

---

## 6. Cobertura territorial

Para cada dataset se evaluó, cuando fue posible:

- número de municipios del universo IOT
- municipios encontrados en la fuente
- municipios homologados
- municipios sin registro
- registros fuera del universo
- registros ambiguos
- registros supramunicipales
- registros departamentales
- registros que no pueden asignarse de forma confiable a un municipio

La ausencia de un registro no se interpreta automáticamente como ausencia del fenómeno.

Debe distinguirse entre:

**cero observado**

y

**ausencia de información o registro**.

Esta distinción es especialmente importante para las fuentes administrativas.

---

## 7. Tratamiento de datos faltantes

Durante la exploración se adoptaron las siguientes reglas generales.

### No se realizó:

- imputación automática de valores
- conversión de `NA` a `0`
- eliminación silenciosa de registros
- sustitución arbitraria de valores
- eliminación automática de valores extremos
- inferencia de ausencia a partir de ausencia de registro

### Sí se realizó:

- identificación explícita de valores faltantes
- distinción entre cero observado y ausencia de registro
- documentación de nulos estructurales
- conservación de valores faltantes cuando la fuente no permite completarlos
- registro de limitaciones de cobertura

Los valores faltantes deben interpretarse de acuerdo con la naturaleza de cada fuente.

---

## 8. Tipología general de las fuentes

A partir de la exploración, las fuentes pueden clasificarse conceptualmente en diferentes grupos.

### A. Fuentes cuantitativas municipales

Fuentes que permiten construir variables directamente comparables entre los municipios del universo IOT.

### B. Fuentes administrativas

Fuentes construidas a partir de registros institucionales, eventos, establecimientos, contratos u otras unidades administrativas que pueden requerir agregación o interpretación antes de convertirse en variables municipales.

### C. Fuentes con limitaciones territoriales

Fuentes que contienen información relevante pero que no permiten construir directamente una variable municipal completa debido a cobertura, unidad de observación o estructura territorial.

### D. Fuentes cualitativas

Fuentes documentales, históricas o de casos que pueden aportar contexto y evidencia cualitativa, pero que no deben convertirse automáticamente en indicadores cuantitativos comparables.

El Dataset 10 corresponde principalmente a esta última categoría.

---

## 9. Reglas de transformación

Las transformaciones realizadas durante la construcción de los datasets deben ser reproducibles y estar documentadas.

Cuando una variable original fue transformada, debe poder identificarse:

- columna original
- transformación aplicada
- columna resultante
- razón de la transformación

No se construyeron indicadores derivados únicamente porque fueran matemáticamente posibles.

Una transformación que implique una decisión analítica debe quedar para revisión metodológica posterior.

---

## 10. Variables categóricas

Las variables categóricas se conservaron en su forma original siempre que fue posible.

No se transformaron automáticamente categorías en valores numéricos.

Por ejemplo, no se asumió que:

- alto = 3
- medio = 2
- bajo = 1

a menos que la propia fuente definiera explícitamente esa escala o existiera una justificación metodológica documentada.

Las variables que pueden parecer ordinales pero cuya estructura no está suficientemente documentada se mantienen como categóricas y quedan sujetas a revisión posterior.

---

## 11. Variables numéricas

Las variables numéricas fueron revisadas considerando:

- tipo de dato
- unidad de medida
- escala
- valores nulos
- valores extremos
- valores imposibles
- duplicados
- consistencia matemática cuando correspondía

Los valores extremos no fueron eliminados únicamente por su magnitud.

Los valores cero explícitos de la fuente se conservaron como cero.

Los valores faltantes no fueron convertidos automáticamente en cero.

---

## 12. Agregación

No se realizó ninguna agregación sin identificar previamente la unidad de observación.

Cuando una fuente contiene múltiples registros asociados a un municipio, primero debe determinarse qué representan esos registros.

Una fuente puede tener estructuras como:

- municipio × categoría
- municipio × año
- municipio × institución
- municipio × evento
- municipio × registro

La agregación solamente se realiza cuando existe una justificación metodológica y la operación conserva el significado de la fuente.

---

## 13. Trazabilidad

Una prioridad transversal del proyecto fue conservar la trazabilidad.

Cuando resulta pertinente, los datasets finales mantienen:

- DIVIPOLA
- municipio homologado
- municipio de la fuente
- fuente original
- unidad de medida
- año
- metadatos
- observaciones metodológicas

Esto permite reconstruir posteriormente cómo se pasó de la fuente RAW a la variable procesada.

## 14. Decisiones metodológicas transversales

### Regla 1 — Universo territorial

El universo territorial del IOT es fijo y está compuesto por 31 municipios del Chocó.

### Regla 2 — DIVIPOLA

La DIVIPOLA es el identificador territorial preferente cuando está disponible.

### Regla 3 — Nombre original

Cuando una fuente proporciona un nombre municipal diferente al nombre homologado, se conserva el nombre original para garantizar trazabilidad.

### Regla 4 — Homologación

La homologación territorial no debe modificar el valor original de la fuente.

### Regla 5 — Ceros y faltantes

Un cero observado no es equivalente a un valor faltante.

### Regla 6 — Agregación

No se agregan registros sin comprender primero la unidad de observación.

### Regla 7 — Construcción de indicadores

La posibilidad matemática de calcular una variable no implica que dicha variable sea metodológicamente válida para el IOT.

### Regla 8 — Fuentes externas

Si una variable requiere información que no se encuentra disponible en RAW, la ausencia debe documentarse antes de incorporarla.

### Regla 9 — No invención

No se deben inventar, imputar o inferir silenciosamente datos que no estén respaldados por la fuente.
