---

## 1. Propósito del proyecto

El Índice de Oportunidades Territoriales (IOT) busca estudiar las condiciones estructurales que enfrentan los habitantes de los municipios del Chocó.

El proyecto parte de una pregunta general:

> ¿Qué condiciones territoriales favorecen o limitan las oportunidades de las personas dependiendo del municipio en el que viven?

El IOT no pretende medir el esfuerzo individual, la calidad moral de los habitantes, el desarrollo integral de un municipio ni determinar causalidad.

Su propósito es identificar y comparar condiciones territoriales relacionadas con:

- conectividad;
- condiciones socioeconómicas;
- ruralidad y estructura territorial;
- acceso/presencia de servicios;
- conectividad física;
- recursos públicos;
- capacidad institucional, cuando exista evidencia suficiente.

El proyecto adopta además una perspectiva de crítica institucional.

Esto significa que los resultados no se limitarán a describir diferencias entre municipios. Cuando los datos permitan identificar patrones potencialmente problemáticos, se formularán preguntas sobre la distribución de recursos, la capacidad institucional, la gestión pública y las desigualdades territoriales.

Sin embargo:

> Una anomalía estadística, una desigualdad o una diferencia en recursos no será presentada automáticamente como evidencia de corrupción.

Las hipótesis de corrupción, captura institucional, mala gestión o uso ineficiente de recursos requerirían evidencia adicional.

# 20. Principio estadístico y narrativo

El proyecto seguirá una regla fundamental:

> **Primero describir. Después comparar. Finalmente preguntar.**

No se buscará que los datos confirmen una acusación previamente formulada.

Cuando aparezca un patrón:

1. se verificará la calidad del dato;
2. se revisará la definición de la variable;
3. se comprobará si el patrón es estadísticamente robusto;
4. se buscarán explicaciones alternativas;
5. se formulará una hipótesis institucional;
6. sólo se afirmará aquello que la evidencia permita sostener.

La crítica institucional será más fuerte si no necesita exagerar los datos.

#### Tabla diagnóstica, no todavía como una conclusión causal. La idea es que podamos ver si los municipios más pobres presentan un patrón territorial distinto.

```python
# ============================================================
# MUNICIPIOS CON MAYOR Y MENOR IPM
# Comparación territorial de pobreza y conectividad
# ============================================================

variables = [
    "municipio",
    "ipm_censo_2018",
    "porcentaje_poblacion_rural",
    "penetracion_internet_fijo_2025_t3_pct",
    "cobertura_movil_3g_4g_pct"
]

df_ipm = df[variables].copy()

# Eliminar municipios sin IPM
df_ipm = df_ipm.dropna(subset=["ipm_censo_2018"])

# Ordenar por IPM
df_ipm = df_ipm.sort_values("ipm_censo_2018", ascending=False)

# Top 10 y bottom 10
top_10_ipm = df_ipm.head(10).copy()
bottom_10_ipm = df_ipm.tail(10).sort_values(
    "ipm_censo_2018",
    ascending=True
).copy()

print("=== 10 MUNICIPIOS CON MAYOR IPM ===")
display(top_10_ipm)

print("\n=== 10 MUNICIPIOS CON MENOR IPM ===")
display(bottom_10_ipm)
```

```python
# ============================================================
# COMPARACIÓN ENTRE LOS DOS GRUPOS
# ============================================================

comparacion_ipm = pd.DataFrame({
    "grupo": ["Mayor IPM (Top 10)", "Menor IPM (Bottom 10)"],
    
    "IPM_promedio": [
        top_10_ipm["ipm_censo_2018"].mean(),
        bottom_10_ipm["ipm_censo_2018"].mean()
    ],
    
    "ruralidad_promedio_pct": [
        top_10_ipm["porcentaje_poblacion_rural"].mean(),
        bottom_10_ipm["porcentaje_poblacion_rural"].mean()
    ],
    
    "internet_fijo_promedio_pct": [
        top_10_ipm["penetracion_internet_fijo_2025_t3_pct"].mean(),
        bottom_10_ipm["penetracion_internet_fijo_2025_t3_pct"].mean()
    ],
    
    "cobertura_movil_promedio_pct": [
        top_10_ipm["cobertura_movil_3g_4g_pct"].mean(),
        bottom_10_ipm["cobertura_movil_3g_4g_pct"].mean()
    ]
})

comparacion_ipm
```
